### 01) WAP to find sum of all the elements in a List.

```python
a=int(input("Enter size"))
li = []
for i in range(a):
    li.append(int(input()))
sum=0
for i in li:
    sum+=i
print(f"Sum = {sum}")
```

### 02) WAP to find largest element in a List.

```python
a=int(input("Enter size"))
li = []
for i in range(a):
    li.append(int(input()))
b=li[0]
for i in li:
    if b<i:
        b=i
print(f"Largest Element = {b}")
```
