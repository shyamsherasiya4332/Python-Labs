### 07) WAP to print sum of series 1 – 2 + 3 – 4 + 5 – 6 + 7 ... n.

```python
n=int(input("Enter Number"))
sum=0
for i in range(1,n+1):
    if i%2==0:
        sum-=i
    else:
        sum+=i
print(sum)
```

### 14) WAP to check whether the given number is Palindrome or not.

```python
n=int(input("Enter Number"))
a=n
rem=0

while a!=0:
    rem=(rem*10)+int(a%10)
    a=a//10
if(n==rem):
    print("Palidrom")
else:
    print("Not Palidrom")
```

### 15) WAP to check whether the given number is an Armstrong Number or not.

```python
n=int(input("Enter Number"))
a=n
b=n
temp=0
count=0

while a!=0:
    count+=1
    a//=10
while b!=0:
    rem=int(b%10)
    temp+=rem**count
    b//=10
if(n==temp):
    print("Armstrong")
else:
    print("Not Armstrong")
```

### 16) WAP to print all the perfect numbers between 1 to n.

```python
n=int(input("Enter Number"))
for i in range(2,n+1):
        sum=0
        for j in range(1,i):
            if i%j==0:
                sum+=j
        if(i==sum):
            print(f"{i}")
```
